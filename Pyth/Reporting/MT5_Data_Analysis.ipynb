## 6. Méthode alternative : Analyse manuelle (pour comparaison)


```python
# Méthode alternative : chargement manuel pour comparaison
print("🔄 Chargement manuel du fichier pour comparaison...")

# Lire toutes les données brutes
df_raw_manual = pd.read_excel(file_path, header=None)

print(f"📊 Fichier chargé manuellement : {file_path}")
print(f"Dimensions : {df_raw_manual.shape[0]} lignes × {df_raw_manual.shape[1]} colonnes")

# Afficher les 10 premières lignes
print("\n🔍 Aperçu des 10 premières lignes :")
display(df_raw_manual.head(10))
```

```python
# Extraction manuelle des informations
def extract_info_value_manual(df, keyword, col_idx=3):
    """Extrait une valeur depuis la section informations."""
    mask = df[0].astype(str).str.contains(keyword, case=False, na=False)
    if mask.any():
        idx = mask.idxmax()
        return df.iloc[idx, col_idx] if col_idx < len(df.columns) else None
    return None

# Extraction des informations clés
info_manual = {
    'expert': extract_info_value_manual(df_raw_manual, 'Expert'),
    'symbole': extract_info_value_manual(df_raw_manual, 'Symbole'),
    'periode': extract_info_value_manual(df_raw_manual, 'Période'),
    'depot_initial': extract_info_value_manual(df_raw_manual, 'Dépôt initial'),
    'profit_total': extract_info_value_manual(df_raw_manual, 'Profit Total Net'),
    'facteur_profit': extract_info_value_manual(df_raw_manual, 'Facteur de profit')
}

print("📋 Informations extraites manuellement :")
for key, value in info_manual.items():
    print(f"  {key.replace('_', ' ').title():.<30} {value}")
```
